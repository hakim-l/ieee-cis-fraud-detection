```python
import os
from pathlib import Path

project_root = Path(os.getcwd()).parent.resolve()
Path(project_root).resolve()
os.chdir(project_root)
```

```python
import pandas as pd
from ieee_cis_fraud_detection import config
```

# Read Data

```python

train_identity= pd.read_csv(
    os.path.join(
        config.RAW_DATA_DIR,
        'train_identity.csv'
    ),
    chunksize= 10000
)

train_transaction = pd.read_csv(
    os.path.join(
        config.RAW_DATA_DIR,
        'train_transaction.csv'
    ),
    chunksize= 10000
)
```

# Partition data

```python
transaction_save_folder= os.path.join(
    config.INTERIM_DATA_DIR,
    'train_transaction_chunks'
)

identity_save_folder= os.path.join(
    config.INTERIM_DATA_DIR,
    'train_identity_chunks'
)

if not os.path.exists(transaction_save_folder):
    os.makedirs(transaction_save_folder)
if not os.path.exists(identity_save_folder):
    os.makedirs(identity_save_folder)
```

**Transaction**

```python
for i, chunk in enumerate(train_transaction):
    chunk.to_parquet(
        os.path.join(
            transaction_save_folder,
            f'train_transaction_chunk_{i}.parquet'
        ),
    )
```

```python
for i, chunk in enumerate(train_identity):
    chunk.to_parquet(
        os.path.join(
            identity_save_folder,
            f'train_identity_chunk_{i}.parquet'
        ),
    )
```
