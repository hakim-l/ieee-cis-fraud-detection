# Read Data

```python
train_transaction_files= [
    os.path.join(
        config.INTERIM_DATA_DIR,
        'train/transactions',
        fn
    )
    for fn in os.listdir(os.path.join(config.INTERIM_DATA_DIR, 'train/transactions'))
]

test_transaction_files= [
    os.path.join(
        config.INTERIM_DATA_DIR,
        'test/transactions',
        fn
    )
    for fn in os.listdir(os.path.join(config.INTERIM_DATA_DIR, 'test/transactions'))
]

identity_files= [
    os.path.join(
        config.INTERIM_DATA_DIR,
        'train/identity',
        fn
    )
    for fn in os.listdir(os.path.join(config.INTERIM_DATA_DIR, 'train/identity'))
]

print(f"Total number of train_transaction files: {len(train_transaction_files)}")
print(f"Total number of identity files: {len(identity_files)}")
```

```python
import pandas as pd

transaction_train= pd.concat(
    [pd.read_parquet(file) for file in train_transaction_files],
    ignore_index=True
)

identity_train= pd.concat(
    [pd.read_parquet(file) for file in identity_files],
    ignore_index=True
)
```

# Merge

```python
identity_train.value_counts('DeviceType')
```

```python
device_info_count= pd.DataFrame(identity_train.value_counts('DeviceInfo'))
device_info_count['cumulative_count'] = device_info_count['count'].cumsum()
device_info_count['cumulative_percentage'] = device_info_count['cumulative_count'] / device_info_count['count'].sum() * 100
device_info_count.reset_index(inplace=True)
device_info_count['category'] = device_info_count['DeviceInfo'].copy()
device_info_count.loc[device_info_count['cumulative_percentage'] > 90, 'category'] = 'Other'
device_info_count.set_index('DeviceInfo', inplace=True)

with open(
        os.path.join(
            config.INTERIM_DATA_DIR,
            'device_info_category_mapping.json'
        ),

        'w'
    ) as f:
    json.dump(
        {
            'column': 'DeviceInfo',
            'mapping': device_info_count.to_dict(orient='index'),
        },
        f
    )

device_info_count
```

```python
def create_categorization(transaction_train, column, save_dir):
    # Implementation for creating categorization
    category_count= pd.DataFrame(transaction_train.value_counts(column))
    category_count['cumulative_count'] = category_count['count'].cumsum()
    category_count['cumulative_percentage'] = category_count['cumulative_count'] / category_count['count'].sum() * 100
    category_count.reset_index(inplace=True)
    category_count['category'] = category_count[column].copy()
    category_count.loc[category_count['cumulative_percentage'] > 90, 'category'] = 'Other'
    category_count.set_index(column, inplace=True)
    with open(
        os.path.join(
            save_dir,
            f'{column}_category_mapping.json'
        ),

        'w'
    ) as f:
        json.dump(
            {
                'column': column,
                'mapping': category_count.to_dict(orient='index'),
            },
            f
        )
    return True
```
