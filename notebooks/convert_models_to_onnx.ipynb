# Load models

**classifier**

```python
import joblib

classifier_model= joblib.load(
    os.path.join(
        MODEL_FOLDER,
        'lgbm_dask_classifier.pkl'
    )
)
```

```python
feature_list= classifier_model.booster_.feature_name()
```

```python
import onnxmltools

initial_type = [('input', onnxmltools.convert.common.data_types.FloatTensorType([None, len(feature_list)]))]
# Convert the LightGBM booster to ONNX
onnx_model = onnxmltools.convert_lightgbm(classifier_model.booster_, initial_types=initial_type)

# Save the ONNX model
onnxmltools.utils.save_model(onnx_model, os.path.join(MODEL_FOLDER, 'lgbm_dask_classifier.onnx'))
```
