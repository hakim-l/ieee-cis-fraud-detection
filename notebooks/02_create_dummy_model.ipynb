```python
train_transaction_files= [
    os.path.join(
        config.INTERIM_DATA_DIR,
        'train_transaction_chunks',
        fn
    )
    for fn in os.listdir(os.path.join(config.INTERIM_DATA_DIR, 'train_transaction_chunks'))
]

identity_files= [
    os.path.join(
        config.INTERIM_DATA_DIR,
        'train_identity_chunks',
        fn
    )
    for fn in os.listdir(os.path.join(config.INTERIM_DATA_DIR, 'train_identity_chunks'))
]

print(f"Total number of train_transaction files: {len(train_transaction_files)}")
print(f"Total number of identity files: {len(identity_files)}")
```

```python
sampled_identities
```

```python
IDENTITY_COLUMNS= [
    'TransactionID',
    'TransactionDT',
    'ProductCD',
]

TARGET_COLUMN= 'isFraud'

CATEGORICAL_COLUMNS= sampled_transactions.select_dtypes(include= 'string[pyarrow]').columns.tolist()
NUMERICAL_COLUMNS= [
    col for col in sampled_transactions.select_dtypes(include= 'number').columns.tolist()
    if col not in IDENTITY_COLUMNS + []
]
```

```python
NUMERICAL_COLUMNS
```

**Initial train to get feature importances**

```python
import numpy as np
sampled_transactions= sampled_transactions.replace(
    {pd.NaT: np.nan}
)
```

```python
import catboost as cb
```

```python
model= cb.CatBoostClassifier(
    iterations= 100,
    learning_rate= 0.1,
    depth= 6,
    eval_metric= 'AUC',
    loss_function= 'Logloss',
    random_seed= 42,
    verbose= 10,
    # cat_features= CATEGORICAL_COLUMNS
)
```

```python
model.fit(
    sampled_transactions.loc[:, NUMERICAL_COLUMNS],
    sampled_transactions[TARGET_COLUMN]
)
```

```python
model.predict(sampled_transactions.loc[:, NUMERICAL_COLUMNS],prediction_type= 'RawFormulaVal')
```

```python
try:
    os.mkdir(
        os.path.join(
            config.MODELS_DIR,
            'eda_catboost_model'
        )
    )
except:
    pass
```

```python
model.save_model(
    os.path.join(
        config.MODELS_DIR,
        'eda_catboost_model',
        'model.onnx'
    ),
    format= 'onnx',
    export_parameters= {
        "onnx_domain": "ai.catboost",
        "prediction_type": "RawFormulaVal",
        'onnx_model_version': 1,
    }
)
```

```python
import onnx

model_onnx= onnx.load(
    os.path.join(
        config.MODELS_DIR,
        'eda_catboost_model',
        'model.onnx'
    )
)
```

```python
model_onnx.graph.input
```

```python
import onnxruntime as ort
import numpy as np

# Load ONNX model
session = ort.InferenceSession(os.path.join(
        config.MODELS_DIR,
        'eda_catboost_model',
        'model.onnx'
    )
    )
```

```python

# Get input and output names
input_name = session.get_inputs()[0].name
output_name = session.get_outputs()[0].name
output_name_2= session.get_outputs()[1].name

print("Input name:", input_name)
print("Output name:", output_name)

# Prepare dummy input (adjust shape & dtype!)
input_shape = session.get_inputs()[0].shape
input_type = session.get_inputs()[0].type

print("INPUT")
print(input_name)
print(input_shape)
print(input_type)

print("OUTPUT")
print(output_name)
print(output_name_2)
```

```python
model_config= {
    "input_name": input_name,
    "label": output_name,
    "probility": output_name_2,
    "columns": NUMERICAL_COLUMNS
}
```

```python
import json
with open(
    os.path.join(
        config.MODELS_DIR,
        'eda_catboost_model',
        'model_config.json'
    ),
    'w'
) as f:
    json.dump(model_config, f, indent= 4)
```

```python


# Example: assume float32 input
dummy_input = sampled_transactions.loc[:,NUMERICAL_COLUMNS].values.astype(np.float32)[:500,:]

# Run inference
outputs = session.run([output_name_2], {input_name: dummy_input})

# print("Output shape:", outputs[0].shape)
```
